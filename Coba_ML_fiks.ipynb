# PREPROCECING

```python
data_path = paths
train_dir = os.path.join(data_path, "Train")
val_dir   = os.path.join(data_path, "Validation")
test_dir  = os.path.join(data_path, "Test")

print(train_dir)
print(val_dir)
print(test_dir)
```

```python
# AUGMENTASI/normlisasi
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# data train
train_transform = transforms.Compose([
    # tinggi,lebar
    transforms.Resize((224,224)),
    # Mengambil bagian acak
    # transforms.RandomCrop((224,224)),
    # hor & rot
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    # ubah color brigh dsb
    transforms.ColorJitter(0.15, 0.15, 0.1),
        # brightness=0.15,
        # contrast=0.15
    # ),
    # Mengubah gambar menjadi tensor PyTorch
    transforms.ToTensor(),
    transforms.Normalize(
        mean=IMAGENET_MEAN,
        std=IMAGENET_STD
    )
])

# data test
test_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=IMAGENET_MEAN,
        std=IMAGENET_STD
    )
])
```

```python
train_data = datasets.ImageFolder(
    train_dir,
    transform=train_transform
)
val_data = datasets.ImageFolder(
    val_dir,
    transform=test_transform
)
test_data = datasets.ImageFolder(
    test_dir,
    transform=test_transform
)

train_loader = DataLoader(
    train_data,
    batch_size=32,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)
val_loader = DataLoader(
    val_data,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
test_loader = DataLoader(
    test_data,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("label:", train_data.classes)
```

```python
img_path = train_data.samples[0][0]
img = Image.open(img_path).convert("RGB")

transform_list ={
    'ori' : transforms.Compose([]),
    'resize' : transforms.Resize((224,224)),
    'crop' : transforms.Compose([
        transforms.Resize((224,224)),
        # transforms.RandomCrop((224,224))

    ]),
    'horizontal flip': transforms.RandomHorizontalFlip(),
    'rotation': transforms.RandomRotation(10),
    'color jitter': transforms.ColorJitter(0.15,0.15,0.10),
        # brightness=0.3,
        # contrast=0.3
    # ),
    "Train Transform Full":train_transform,

    "Test Transform":test_transform
}

# tampilkan
plt.figure(figsize=(16,10))

for i, (name, transform) in enumerate(transform_list.items()):

    transformed = transform(img)


    if hasattr(transformed, "shape"):
        transformed = transformed.permute(1,2,0).numpy()
        if transformed.min() < 0:
            transformed = transformed * IMAGENET_STD + IMAGENET_MEAN

        transformed = transformed.clip(0,1)

    plt.subplot(2,4,i+1)
    plt.imshow(transformed)
    plt.title(name)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def model_tuning(mode):
  model = models.resnet50(
        weights=models.ResNet50_Weights.IMAGENET1K_V2
    )
  # FEATURE EXTRACTION vs FINE TUNING
  if mode == "fe":
    for param in model.parameters():
        param.requires_grad = False
  elif mode == "ft":
        for param in model.parameters():
            param.requires_grad = True

  jumlah_fitur = model.fc.in_features
  model.fc = nn.Linear(jumlah_fitur, 2)
  return model
```

```python
# fitur ekstraksi
model_fe = model_tuning("fe")
# model_fe = model_fe.cuda()
model_fe = model_fe
model_fe
```

### training

```python
# loss
criterion = nn.CrossEntropyLoss()
# ganti model_ft atau model_fe.fc
optimizer = optim.Adam(
    model_fe.fc.parameters(),
    lr=1e-3
)
```

```python
EPOCH = 50

train_loss = []
val_loss = []
train_acc = []
val_acc = []


best_epoch = 0
best_train_acc = 0
best_val_acc = 0
best_train_loss = 0
best_val_loss = 0
best_acc_gap = 0
best_loss_gap = 0

best_score_main = -float('inf')
best_score_fallback = -float('inf')
best_model_found = False
best_model_type = "NONE"

patience = 10
counter = 0

for epoch in range (EPOCH):
  model_fe.train()

  run_loss = 0
  train_preds = []
  train_labels = []
# TRAIN
  for images, labels in train_loader:
    images = images.to(DEVICE)
    labels = labels.to(DEVICE)
    optimizer.zero_grad()

    output = model_fe(images)
    loss = criterion(output, labels)
    loss.backward()
    optimizer.step()

    run_loss += loss.item()
    preds = torch.argmax(output, dim=1)

    train_preds.extend(preds.cpu().numpy())
    train_labels.extend(labels.cpu().numpy())

  train_los = run_loss / len(train_loader)
  train_ac = accuracy_score(
      train_labels,
      train_preds
  )

  # VAL
  model_fe.eval()
  run_loss = 0

  val_preds = []
  val_labels = []

  with torch.no_grad():
    for images, labels in val_loader:
      images = images.to(DEVICE)
      labels = labels.to(DEVICE)
      output = model_fe(images)

      loss = criterion(output, labels)
      run_loss += loss.item()

      preds = torch.argmax(output, dim=1)

      val_preds.extend(preds.cpu().numpy())
      val_labels.extend(labels.cpu().numpy())

    val_los = run_loss / len(val_loader)
    val_ac = accuracy_score(
        val_labels,
        val_preds
    )

    # save hist
    train_loss.append(train_los)
    val_loss.append(val_los)
    train_acc.append(train_ac*100)
    val_acc.append(val_ac*100)

    # best model
    # if val_ac > best_acc:
      # best_acc = val_ac
      # torch.save(
          # model_fe.state_dict(),
          # "best_resnet34_fe.pth"
      # )
    acc_gap = abs(train_ac - val_ac) * 100
    loss_gap = (val_los - train_los)
    loss_score = 1/(1+val_los)
    score = (
        0.7 *val_ac + 0.3 * loss_score
    )
    fallback = (
        0.6 * val_ac
        + 0.2 * loss_score
        + 0.1 * (1 - min(acc_gap/20,1))
        + 0.1 * (1 - min(loss_gap/0.20,1))
    )
    if (acc_gap <= 20) and (loss_gap <= 0.20):
      if  (score > best_score_main) :
        best_model_found = True
        best_model_type = "MAIN"
        best_score_main = score
        best_epoch = epoch + 1
        best_train_acc = (train_ac * 100)
        best_val_acc = (val_ac * 100)
        best_train_loss = (train_los)
        best_val_loss = (val_los)
        best_acc_gap = (acc_gap)
        best_loss_gap = (loss_gap)
        counter = 0
        torch.save(
          model_fe.state_dict(),
          "best_resnet34_fe.pth"
        )
        
        print(
            f"Best Model Saved | "
            f"Score={score:.4f} | "
            f"Val Acc={val_ac*100:.2f}% | "
            f"Val Loss={val_los:.4f}"
        )
      else :
              counter += 1
    
    else:
      if (not best_model_found) and (fallback > best_score_fallback):
        best_model_type = "FALLBACK"
        best_score_fallback = fallback
        best_epoch = epoch + 1
        best_train_acc = train_ac * 100
        best_val_acc = val_ac * 100
        best_train_loss = train_los
        best_val_loss = val_los
        best_acc_gap = acc_gap
        best_loss_gap = loss_gap
        counter = 0
        torch.save(
            model_fe.state_dict(),
            "best_resnet34_fe.pth"
        )
        
      else:
        counter += 1
      
    # live graph
    clear_output(wait = True)
    fig, ax = plt.subplots(1,2, figsize=(14,5))

    # LOSS
    ax[0].plot(train_loss, marker='o', label='Train')
    ax[0].plot(val_loss, marker='o', label='Validation')

    ax[0].set_title("Loss")
    ax[0].set_xlabel("Epoch")
    ax[0].legend()

    # ACCURACY
    ax[1].plot(train_acc, marker='o', label='Train')
    ax[1].plot(val_acc, marker='o', label='Validation')

    ax[1].set_title("Accuracy")
    ax[1].set_xlabel("Epoch")
    ax[1].legend()

    plt.suptitle(
        f"Epoch {epoch+1}/{EPOCH} | "
        f"Val Acc: {val_ac*100:.2f}%"
    )

    plt.tight_layout()
    plt.show()

    print(
        f"Epoch [{epoch+1}/{EPOCH}] "
        f"| Train Loss: {train_los:.4f} "
        f"| Val Loss: {val_los:.4f} "
        f"| Train Acc: {train_ac*100:.2f}% "
        f"| Val Acc: {val_ac*100:.2f}%"
        f"| Acc Gap: {acc_gap:.2f}% "
        f"| Loss Gap: {loss_gap:.4f} "
        f"| Score: {score:.4f} "
        f"| Counter: {counter}/{patience}"
    )
    
    print(
        f"\nCURRENT BEST MODEL "
        f"| Type: {best_model_type} "
        f"| Epoch: {best_epoch} "
        f"| Val Acc: {best_val_acc:.2f}% "
        f"| Acc Gap: {best_acc_gap:.2f}% "
        f"| Loss Gap: {best_loss_gap:.4f}"
    )
    if best_model_type == "MAIN":
        print(f"Final Score     : {best_score_main:.4f}")
    else:
        print(f"Final Score     : {best_score_fallback:.4f}")
    if counter >= patience:
      print(f"\n Early Stopping pada Epoch {epoch+1}")
      break
    
    
if best_model_type == "MAIN":
    final_score = best_score_main
else:
    final_score = best_score_fallback
print("\n" + "="*50)
print("BEST MODEL")
print("="*50)

print(f"Model Type     : {best_model_type}")
print(f"Epoch           : {best_epoch}")


print(f"Train Accuracy  : {best_train_acc:.2f}%")
print(f"Validation Acc  : {best_val_acc:.2f}%")
print(f"Train Loss      : {best_train_loss:.4f}")
print(f"Validation Loss : {best_val_loss:.4f}")
print(f"Accuracy Gap    : {best_acc_gap:.2f}%")
print(f"Loss Gap        : {best_loss_gap:.4f}")
print(f"Final Score     : {final_score:.4f}")
print("="*50)
```

```python
EPOCH = 100

train_loss = []
val_loss = []
train_acc = []
val_acc = []


best_epoch = 0
best_train_acc = 0
best_val_acc = 0
best_train_loss = 0
best_val_loss = 0
best_acc_gap = 0
best_loss_gap = 0

best_score_main = -float('inf')
best_score_fallback = -float('inf')
best_model_found = False
best_model_type = "NONE"

patience = 10
counter = 0

for epoch in range (EPOCH):
  model_efficientnet_ft.train()

  run_loss = 0
  train_preds = []
  train_labels = []
# TRAIN
  for images, labels in train_loader_bo:
    images = images.to(DEVICE)
    labels = labels.to(DEVICE)
    optimizer.zero_grad()

    output = model_efficientnet_ft(images)
    loss = criterion(output, labels)
    loss.backward()
    optimizer.step()

    run_loss += loss.item()
    preds = torch.argmax(output, dim=1)

    train_preds.extend(preds.cpu().numpy())
    train_labels.extend(labels.cpu().numpy())

  train_los = run_loss / len(train_loader_bo)
  train_ac = accuracy_score(
      train_labels,
      train_preds
  )

  # VAL
  model_efficientnet_ft.eval()
  run_loss = 0

  val_preds = []
  val_labels = []

  with torch.no_grad():
    for images, labels in val_loader_bo:
      images = images.to(DEVICE)
      labels = labels.to(DEVICE)
      output = model_efficientnet_ft(images)

      loss = criterion(output, labels)
      run_loss += loss.item()

      preds = torch.argmax(output, dim=1)

      val_preds.extend(preds.cpu().numpy())
      val_labels.extend(labels.cpu().numpy())

    val_los = run_loss / len(val_loader_bo)
    val_ac = accuracy_score(
        val_labels,
        val_preds
    )

    # save hist
    train_loss.append(train_los)
    val_loss.append(val_los)
    train_acc.append(train_ac*100)
    val_acc.append(val_ac*100)

    # best model
    # if val_ac > best_acc:
      # best_acc = val_ac
      # torch.save(
          # model_efficientnet_ft.state_dict(),
          # "best_efficientnet_ft.pth"
      # )
    acc_gap = abs(train_ac - val_ac) * 100
    loss_gap = abs(val_los - train_los)
    loss_score = 1/(1+val_los)
    score = (
        0.7 *val_ac + 0.3 * loss_score
    )
    fallback = (
        0.6 * val_ac
        + 0.2 * loss_score
        + 0.1 * (1 - min(acc_gap/20,1))
        + 0.1 * (1 - min(loss_gap/0.20,1))
    )
    if (acc_gap <= 20) and (loss_gap <= 0.20):
      if  (score > best_score_main) :
        best_model_found = True
        best_model_type = "MAIN"
        best_score_main = score
        best_epoch = epoch + 1
        best_train_acc = (train_ac * 100)
        best_val_acc = (val_ac * 100)
        best_train_loss = (train_los)
        best_val_loss = (val_los)
        best_acc_gap = (acc_gap)
        best_loss_gap = (loss_gap)
        counter = 0
        torch.save(
          model_efficientnet_ft.state_dict(),
          "best_efficientnet_ft.pth"
        )
        
        print(
            f"Best Model Saved | "
            f"Score={score:.4f} | "
            f"Val Acc={val_ac*100:.2f}% | "
            f"Val Loss={val_los:.4f}"
        )
      else :
              counter += 1
    
    else:
      if (not best_model_found) and (fallback > best_score_fallback):
        best_model_type = "FALLBACK"
        best_score_fallback = fallback
        best_epoch = epoch + 1
        best_train_acc = train_ac * 100
        best_val_acc = val_ac * 100
        best_train_loss = train_los
        best_val_loss = val_los
        best_acc_gap = acc_gap
        best_loss_gap = loss_gap
        counter = 0
        torch.save(
            model_efficientnet_ft.state_dict(),
            "best_efficientnet_ft.pth"
        )
        
      else:
        counter += 1
      
    # live graph
    clear_output(wait = True)
    fig, ax = plt.subplots(1,2, figsize=(14,5))

    # LOSS
    ax[0].plot(train_loss, marker='o', label='Train')
    ax[0].plot(val_loss, marker='o', label='Validation')

    ax[0].set_title("Loss")
    ax[0].set_xlabel("Epoch")
    ax[0].legend()

    # ACCURACY
    ax[1].plot(train_acc, marker='o', label='Train')
    ax[1].plot(val_acc, marker='o', label='Validation')

    ax[1].set_title("Accuracy")
    ax[1].set_xlabel("Epoch")
    ax[1].legend()

    plt.suptitle(
        f"Epoch {epoch+1}/{EPOCH} | "
        f"Val Acc: {val_ac*100:.2f}%"
    )

    plt.tight_layout()
    plt.show()

    print(
        f"Epoch [{epoch+1}/{EPOCH}] "
        f"| Train Loss: {train_los:.4f} "
        f"| Val Loss: {val_los:.4f} "
        f"| Train Acc: {train_ac*100:.2f}% "
        f"| Val Acc: {val_ac*100:.2f}%"
        f"| Acc Gap: {acc_gap:.2f}% "
        f"| Loss Gap: {loss_gap:.4f} "
        f"| Score: {score:.4f} "
        f"| Counter: {counter}/{patience}"
    )
    
    print(
        f"\nCURRENT BEST MODEL "
        f"| Type: {best_model_type} "
        f"| Epoch: {best_epoch} "
        f"| Val Acc: {best_val_acc:.2f}% "
        f"| Acc Gap: {best_acc_gap:.2f}% "
        f"| Loss Gap: {best_loss_gap:.4f}"
    )
    if best_model_type == "MAIN":
        print(f"Final Score     : {best_score_main:.4f}")
    else:
        print(f"Final Score     : {best_score_fallback:.4f}")
    if counter >= patience:
      print(f"\n Early Stopping pada Epoch {epoch+1}")
      break
    
    
if best_model_type == "MAIN":
    final_score = best_score_main
else:
    final_score = best_score_fallback
print("\n" + "="*50)
print("BEST MODEL")
print("="*50)

print(f"Model Type     : {best_model_type}")
print(f"Epoch           : {best_epoch}")


print(f"Train Accuracy  : {best_train_acc:.2f}%")
print(f"Validation Acc  : {best_val_acc:.2f}%")
print(f"Train Loss      : {best_train_loss:.4f}")
print(f"Validation Loss : {best_val_loss:.4f}")
print(f"Accuracy Gap    : {best_acc_gap:.2f}%")
print(f"Loss Gap        : {best_loss_gap:.4f}")
print(f"Final Score     : {final_score:.4f}")
print("="*50)
```

```python
# MODEL LIST
model_dict = {
    "ResNet50 FE": model_fe,
    "ResNet50 FT": model_ft,
    "EfficientNet-B0 FE": model_efficientnet_fe,
    "EfficientNet-B0 FT": model_efficientnet_ft
}


# TEST IMAGE
image_paths = [x[0] for x in test_data.samples]
image_names = [os.path.basename(x)for x in image_paths]

image_dict = dict(
    zip(
        image_names,
        image_paths
    )
)


# SHOW GRADCAM
def show_gradcam(model_name, selected_image):
    model = model_dict[model_name]
    model.eval()
    # GRAD FE
    if "FE" in model_name:
        if "ResNet" in model_name:
            for p in model.layer4.parameters():
                p.requires_grad = True
        else:
            for p in model.features[-1].parameters():
                p.requires_grad = True

    image_path = image_dict[selected_image]
    gradients = []
    activations = []
    
    # HOOK
    def forward_hook(module, input, output):
        activations.append(output)
        if output.requires_grad:
            def save_gradients(grad):
                gradients.append(grad)
            output.register_hook(save_gradients)
            
    # TARGET LAYER
    if "ResNet" in model_name:
        target_layer = model.layer4[-1]
    else:
        target_layer = model.features[-1]
    handle = target_layer.register_forward_hook(forward_hook)

    # IMAGE
    img = Image.open(image_path).convert("RGB")
    img_np = np.array(img)
    transform = transforms.Compose([
        transforms.Resize((224,224)),
        transforms.ToTensor(),
        transforms.Normalize(
            mean=[0.485,0.456,0.406],
            std=[0.229,0.224,0.225]
        )
    ])
    input_tensor = transform(img).unsqueeze(0).to(DEVICE)
    input_tensor.requires_grad_(True)
    # FORWARD
    output = model(input_tensor)
    pred_class = torch.argmax(output,dim=1).item()
    confidence = torch.softmax(output,dim=1)[0,pred_class].item()
    
    # BACKWARD
    model.zero_grad()
    output[0,pred_class].backward()
    grads = gradients[0][0].detach().cpu().numpy()
    fmap = activations[0][0].detach().cpu().numpy()
    weights = np.mean(grads,axis=(1,2))
    cam = np.zeros(fmap.shape[1:],dtype=np.float32)
    
    for i,w in enumerate(weights):
        cam += w * fmap[i]
    cam = np.maximum(cam,0)
    cam = cv2.resize(cam,(224,224))
    cam -= cam.min()
    if cam.max() > 0:
        cam /= cam.max()
        
    # HEATMAP
    heatmap = cv2.applyColorMap(np.uint8(255 * cam),cv2.COLORMAP_JET)
    heatmap = cv2.cvtColor(heatmap,cv2.COLOR_BGR2RGB)
    overlay = (
        0.4 * heatmap +
        0.6 * cv2.resize(img_np,(224,224))
        )
    overlay = overlay.astype(np.uint8)


    # CLASS
    class_name = train_data.classes[pred_class]

    # PLOT
    fig, ax = plt.subplots(1,3,figsize=(16,5))
    ax[0].imshow(img_np)
    ax[0].set_title("Original")
    ax[0].axis("off")

    ax[1].imshow(cam,cmap="jet")
    ax[1].set_title("Grad-CAM")
    ax[1].axis("off")

    ax[2].imshow(overlay)
    ax[2].set_title("Overlay")
    ax[2].axis("off")

    plt.suptitle(
        f"{model_name}\n"
        f"Prediction : {class_name}\n"
        f"Confidence : {confidence*100:.2f}%",

        fontsize=14,
        fontweight="bold"
    )
    plt.tight_layout()
    plt.show()
    handle.remove()

# DROPDOWN
widgets.interact(
    show_gradcam,
    model_name=widgets.Dropdown(options=list(model_dict.keys()),description="Model:"),
    selected_image=widgets.Dropdown(
        options=image_names,
        description="Image:"
    )
)
```

```python
# CLASS NAME
CLASS_NAMES = [
    "Defective",
    "Non Defective"
]

def predict_with_confidence(img_pil):
    fig, axes = plt.subplots(1,len(MODELS_DICT)+1,figsize=(40,5))
    
    # INPUT IMAGE
    axes[0].imshow(img_pil)
    axes[0].set_title("Input Image",fontsize=12,fontweight="bold")
    axes[0].axis("off")
    
    # MODEL LOOPING
    for ax, (name, info) in zip(axes[1:],MODELS_DICT.items()):
        try:
            mode = info[0]
            
            # CNN
            if mode == "cnn":
                model = info[1]
                tfm = info[2]
                model.eval()
                x = tfm(img_pil).unsqueeze(0).to(DEVICE)
                with torch.no_grad():
                    probs = F.softmax(model(x),dim=1)[0].cpu().numpy()
                    
            # SVM RESNET
            elif mode == "svm_resnet":
                x = test_transform(img_pil).unsqueeze(0).to(DEVICE)
                with torch.no_grad():
                    feat = feature_model(x)
                feat = feat.cpu().numpy()
                feat = scaler.transform(feat)
                feat = pca.transform(feat)
                probs = svm.predict_proba(feat)[0]

            
            # RF RESNET
            elif mode == "rf_resnet":
                x = test_transform(img_pil).unsqueeze(0).to(DEVICE)
                with torch.no_grad():
                    feat = feature_model(x)
                feat = feat.cpu().numpy()
                feat = scaler.transform(feat)
                feat = pca.transform(feat)
                probs = rf.predict_proba(feat)[0]

            # SVM EFFICIENTNET            
            elif mode == "svm_eff":
                x = test_transform_bo(img_pil).unsqueeze(0).to(DEVICE)
                with torch.no_grad():
                    feat = feature_model_bo(x)
                feat = feat.cpu().numpy()
                feat = scaler_bo.transform(feat)
                feat = pca_bo.transform(feat)

                probs = svm_bo.predict_proba(feat)[0]

            # RF EFFICIENTNET            
            elif mode == "rf_eff":
                x = test_transform_bo(img_pil).unsqueeze(0).to(DEVICE)
                with torch.no_grad():
                    feat = feature_model_bo(x)
                feat = feat.cpu().numpy()
                feat = scaler_bo.transform(feat)
                feat = pca_bo.transform(feat)
                probs = rf_bo.predict_proba(feat)[0]

            
            # SORT
            idx = np.argsort(probs)[::-1]
            probs_sorted = probs[idx]
            names_sorted = [
                CLASS_NAMES[i]
                for i in idx
            ]

            colors = [
                "#2196F3"
                if i == 0
                else "#90CAF9"
                for i in range(len(probs_sorted))
            ]

            bars = ax.barh(
                range(len(probs_sorted)),
                probs_sorted,
                color=colors
            )

            ax.set_yticks(range(len(probs_sorted)))
            ax.set_yticklabels(names_sorted,fontsize=8)
            ax.set_xlim(0,1)
            ax.set_title(f"{name}\nTop: {names_sorted[0]}",fontsize=9,fontweight="bold")
            ax.invert_yaxis()

            for bar,p in zip(bars,probs_sorted):
                ax.text(
                    p+0.01,
                    bar.get_y()+bar.get_height()/2,
                    f"{p*100:.1f}%",
                    va="center",
                    fontsize=8
                )

        except Exception as e:
            print(f"\nERROR pada {name}")
            print(e)
            
            ax.text(
                0.5,
                0.5,
                str(e),
                ha="center",
                va="center",
                wrap=True
            )
            ax.set_title(name)
            ax.axis("off")

    plt.tight_layout()
    plt.show()

# TEST IMAGE LIST
image_paths = [
    x[0]
    for x in test_data.samples
]
img_dropdown = widgets.Dropdown(
    options=image_paths,
    description="Image:",
    layout=widgets.Layout(width="80%")
)

# INTERACTIVE
def show_conf(img_path):
    img_pil = Image.open(img_path).convert("RGB")
    predict_with_confidence(img_pil)

widgets.interact(
    show_conf,
    img_path=img_dropdown
)
```

```python
def get_all_probs(img_pil):
    probs_list = []

    # ResNet FE
    x = test_transform(img_pil).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        probs = F.softmax(
            model_fe(x),
            dim=1
        )[0].cpu().numpy()

    probs_list.append(probs)

    # ResNet FT
    with torch.no_grad():
        probs = F.softmax(
            model_ft(x),
            dim=1
        )[0].cpu().numpy()

    probs_list.append(probs)

    # EfficientNet FE
    x_bo = test_transform_bo(img_pil).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        probs = F.softmax(
            model_efficientnet_fe(x_bo),
            dim=1
        )[0].cpu().numpy()
    probs_list.append(probs)

    # EfficientNet FT
    with torch.no_grad():
        probs = F.softmax(
            model_efficientnet_ft(x_bo),
            dim=1
        )[0].cpu().numpy()

    probs_list.append(probs)

    # ResNet Feature
    with torch.no_grad():
        feat_resnet = feature_model(x)
    feat_resnet = feat_resnet.cpu().numpy()
    feat_resnet = scaler.transform(feat_resnet)
    feat_resnet = pca.transform(feat_resnet)
    
    # SVM ResNet
    probs_list.append(svm.predict_proba(feat_resnet)[0])

    # RF ResNet
    probs_list.append(rf.predict_proba(feat_resnet)[0])

    # EfficientNet Feature
    with torch.no_grad():
        feat_eff = feature_model_bo(x_bo)
    feat_eff = feat_eff.cpu().numpy()
    feat_eff = scaler_bo.transform(feat_eff)

    feat_eff = pca_bo.transform(feat_eff)

    # SVM EfficientNet
    probs_list.append(svm_bo.predict_proba(feat_eff)[0])

    # RF EfficientNet
    probs_list.append(rf_bo.predict_proba(feat_eff)[0])
    return np.array(probs_list)
```

```python
y_true_soft = []
y_pred_soft = []

for img_path, label in test_data.samples:

    img_pil = Image.open(img_path).convert("RGB")
    probs_all = get_all_probs(img_pil)
    mean_probs = np.mean(probs_all,axis=0)
    pred = np.argmax(mean_probs)
    y_true_soft.append(label)
    y_pred_soft.append(pred)

y_true_soft = np.array(y_true_soft)
y_pred_soft = np.array(y_pred_soft)
```

```python
from sklearn.metrics import accuracy_score

soft_acc = accuracy_score(
    y_true_soft,
    y_pred_soft
)

print(
    f"Soft Voting Accuracy : "
    f"{soft_acc*100:.2f}%"
)
```

```python
def evaluate_soft_voting(dataset):
    y_true = []
    y_pred = []

    for img_path, label in dataset.samples:

        img_pil = Image.open(img_path).convert("RGB")
        probs_all = get_all_probs(img_pil)

        mean_probs = np.mean(
            probs_all,
            axis=0
        )
        pred = np.argmax(mean_probs)

        y_true.append(label)
        y_pred.append(pred)

    acc = accuracy_score(y_true,y_pred)
    return acc


train_acc_soft = evaluate_soft_voting(train_data)
test_acc_soft = evaluate_soft_voting(test_data)
gap_acc_soft = abs(train_acc_soft -test_acc_soft) * 100

print(f"Train Acc : {train_acc_soft*100:.2f}%")
print(f"Test Acc  : {test_acc_soft*100:.2f}%")
print(f"Gap Acc   : {gap_acc_soft:.2f}%")
```
